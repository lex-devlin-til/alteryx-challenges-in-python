### Difference between Series and DataFrame

- Series is a 1 Dimensional labeled array. Example: A column in Excel or a column in a DataFrame
https://pandas.pydata.org/docs/reference/series.html

- DataFrame is a 2-Dimensional labeled data structure. Example: A table with rows and columns
https://pandas.pydata.org/docs/reference/frame.html

```python
import pandas as pd
```

```python
#Example of creating a Pandas Series
#example_list is a list of data values
example_list = ["DS1", "DS2", "DS3", "DS4", "DS5"]
index = [0, 1, 2, 3, 4]
s = pd.Series(data=example_list, index=index)
s
```

```python
s[0]
```

```python
#Example of creating a Pandas DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'Salary': [50000, 60000, 70000]
}

df = pd.DataFrame(data)
print(df)
```

```python
#Return a Series representing the 'Name' column
df['Name']
```

```python
df['Name'].shape
```

```python
#Return a DataFrame with only the 'Name' and 'Age' columns
df[['Name', 'Age']]
```

```python
df[['Name']]
```

```python
df[['Name']].shape
```

```python

df = pd.DataFrame({'A':[1,2,3], 'B':[4,5,6]})
s = df['A']

print("Mean from a Dataframe:")   # works → returns mean per column
print(df.mean())

print("\nMean from a Series for column A only:")    # works → returns scalar
print(s.mean())
```

### Indexing and Selection
Loc is based on the **Label** for row name, column name
- loc[] allows the input is a single label, a ilst, a slice object
- If selecting one row df.loc[<index_name>]
- selecting multiple rows df.loc[[<list_of_index>]]
- selecting multiple rows and specified columns: df.loc[[<list_of_index or a_range_if_idx_is_integer>],[<list_of_columns>]]
- selecting a subset of data with condition: df.loc[df['column_name']>0,[<list_of_columns>]]
- for mult-conditions: df.loc[(df['column_name']>0) & (df['column_name']=='abc'),[<list_of_columns>]]

```python
excel_input.head()
```

```python
#Set index is the Order ID column
excel_input.set_index('Order ID', inplace=True)
```

```python
excel_input.head()
```

```python
#Selecting row with index label 1 => showing all columns for that row
excel_input.loc['CA-2016-152156']
```

```python
#Select multiple rows by passing a list of index
excel_input.loc[['CA-2016-138688','CA-2016-152156']]
```

```python
#Select a range of rows from 0 to 2 for 3 columns: 'Order ID', 'Sales', 'Profit'
excel_input.loc[['CA-2016-152156','CA-2016-152156','CA-2016-138688'],['Sales','Profit']]
```

```python
#Can be used to filter data with condition
excel_input.loc[excel_input['Profit']>0,['Sales','Profit']]
```

```python
#Multiple conditions and select specific columns
excel_input.loc[(excel_input['Profit']>0) & (excel_input['Sales']>500),['Sales','Profit']]
```

iloc is based on the **Integer Position** from **0 to length-1**, but also a boolean array
- With an integer input to select 1 row
- To select multiple rows: df.iloc[[<list_of_integer_position_indx>]] or df.iloc[<a_range_of_indx>]
- can index both axes: df.iloc[0,1] or df.iloc[[0,2],[1,5]] or df.iloc[[0:3],[2:6]]
- CANNOT filter the data with condition because iloc is integer position base, not for boolean conditions

```python
excel_input.iloc[[0,4,5]]
```

```python
#Select a range for rows and columns
#Remember that the ending index is EXCLUSIVE
excel_input.iloc[0:2,1:3]
```

```python
excel_input.iloc[:10, [0,2,3,5]]
```

### Sorting
- df.sort_values(by=[<column_name or list_of_columns>], ascending=True or False, na_position='first' or 'last')
For sorting the dataframe by a column or list of columns in ascending or descending; can also set the NA in the first or last row.

df.sort_index(ascending=True or False, na_position='last' or 'first')

```python
#Sorting the dataframe by 'Sales' in ascending order
excel_input.sort_values(by='Sales', ascending=True).head(5)
```

```python
#Sorting the dataframe by 2 columns: 'Country' in ascending order and 'Sales' in descending order
#Using loc to select specific columns after sorting
excel_input.sort_values(by=['Country','Sales'], ascending=[True, False]).head(5).loc[:,['Order ID','Country','Sales']]
```
